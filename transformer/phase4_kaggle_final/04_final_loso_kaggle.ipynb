# Transformer final training (Stage D) + determinism rerun + 6-fold LOSO CV

Runs `transformer/PLAN.md` Phase T4: trains the Phase-T3 winner (`transformer_searched`)
and the pre-registered `transformer_default`, 5 seeds each, with **test set03 evaluated
here and nowhere else in the entire project**. Also runs a seed-42
determinism-of-record rerun (must reproduce its own numbers -- the Issue-2 "seed 42
reproduces 0.9131" precedent applied to the transformer) and the 6-fold
leave-one-set-out CV (Issue-5 protocol) on the winner config.

**Winner config** (from `phase3_search_review/03_search_summary.md`, Phase T3, reviewed
and confirmed by the user before this notebook was built):
`d128_ff512_L4_last_spe__adam_lr1e-03_plateau_do0.1_wd1e-05` -- d_model=128,
num_layers=4, dim_ff=512, pool="last", pos="sin", Adam lr=1e-3/plateau/dropout=0.1/wd=1e-5.
Val AUC (5-seed) 0.9789 +/- 0.0038, beating `transformer_default`'s 0.9629 +/- 0.0056
(which ranked 66th/78 in the seed-42 architecture pool). The literal config dict below
was copied from that config's raw `seed42.json` in `phase2_kaggle_search/runs_search/`,
not hand-parsed from the cfg_id string.

```python
# KEEP IN SYNC WITH transformer/phase1_setup/00_transformer_model.py -- edit there
# first, then regenerate this notebook (see gen_final_nb.py). Do not hand-edit this cell.
model_code = '"""00_transformer_model.py — TransformerIntentPredictor.\n\nSingle source of truth for the transformer architecture (transformer/PLAN.md #3).\nThe Kaggle notebooks (phase2_kaggle_search/03_search_kaggle.ipynb,\nphase4_kaggle_final/04_final_loso_kaggle.ipynb) embed this class verbatim under a\n"KEEP IN SYNC WITH 00_transformer_model.py" banner — edit here first, then copy down.\n\nForward contract matches pipeline/03_bilstm_model.py::BiLSTMIntentPredictor:\n    input  (B, 16, 5) float32, already z-scored by the caller\n    output (B, 1) float32 raw logits (sigmoid applied by the caller)\nNo required constructor args (defaults = transformer_default, PLAN.md #3).\n"""\nimport math\n\nimport torch\nimport torch.nn as nn\n\n\ndef _sinusoidal_pe(length, d_model):\n    pe = torch.zeros(length, d_model)\n    position = torch.arange(length, dtype=torch.float32).unsqueeze(1)\n    div_term = torch.exp(torch.arange(0, d_model, 2, dtype=torch.float32)\n                         * (-math.log(10000.0) / d_model))\n    pe[:, 0::2] = torch.sin(position * div_term)\n    pe[:, 1::2] = torch.cos(position * div_term[: pe[:, 1::2].shape[1]])\n    return pe.unsqueeze(0)\n\n\nclass TransformerIntentPredictor(nn.Module):\n    def __init__(self, input_dim=5, d_model=128, nhead=4, num_layers=2, dim_ff=None,\n                 dropout=0.1, pool="cls", pos="learned", seq_len=16, norm_first=True):\n        super().__init__()\n        if d_model % nhead != 0:\n            raise ValueError(f"d_model={d_model} must be divisible by nhead={nhead}")\n        if pool not in ("cls", "mean", "last"):\n            raise ValueError(f"pool must be one of cls/mean/last, got {pool!r}")\n        if pos not in ("learned", "sin"):\n            raise ValueError(f"pos must be one of learned/sin, got {pos!r}")\n\n        dim_ff = dim_ff or 2 * d_model\n        self.pool = pool\n        self.use_cls = pool == "cls"\n        self.seq_len = seq_len\n\n        self.input_proj = nn.Linear(input_dim, d_model)\n\n        pe_len = seq_len + 1 if self.use_cls else seq_len\n        if self.use_cls:\n            self.cls_token = nn.Parameter(torch.randn(1, 1, d_model) * 0.02)\n        if pos == "learned":\n            self.pos_embed = nn.Parameter(torch.randn(1, pe_len, d_model) * 0.02)\n        else:\n            self.register_buffer("pos_embed", _sinusoidal_pe(pe_len, d_model), persistent=False)\n\n        self.embed_drop = nn.Dropout(dropout)\n\n        if num_layers > 0:\n            layer = nn.TransformerEncoderLayer(\n                d_model, nhead, dim_feedforward=dim_ff, dropout=dropout,\n                activation="gelu", batch_first=True, norm_first=norm_first,\n            )\n            self.encoder = nn.TransformerEncoder(\n                layer, num_layers=num_layers, norm=nn.LayerNorm(d_model),\n                enable_nested_tensor=False,\n            )\n        else:\n            # Gate-1 linear-probe floor: skip the encoder entirely (PLAN.md Phase T1 Gate 1).\n            self.encoder = None\n\n        self.head_drop = nn.Dropout(dropout)\n        self.head = nn.Linear(d_model, 1)\n\n    def forward(self, x):\n        B = x.shape[0]\n        h = self.input_proj(x)\n        if self.use_cls:\n            cls = self.cls_token.expand(B, -1, -1)\n            h = torch.cat([cls, h], dim=1)\n        h = h + self.pos_embed[:, : h.shape[1], :]\n        h = self.embed_drop(h)\n        if self.encoder is not None:\n            h = self.encoder(h)\n\n        if self.pool == "cls":\n            pooled = h[:, 0, :]\n        elif self.pool == "last":\n            pooled = h[:, -1, :]\n        else:  # mean — exclude the CLS token if present\n            seq = h[:, 1:, :] if self.use_cls else h\n            pooled = seq.mean(dim=1)\n\n        pooled = self.head_drop(pooled)\n        return self.head(pooled)\n\n\ndef count_params(model):\n    return sum(p.numel() for p in model.parameters())\n\n\nif __name__ == "__main__":\n    torch.manual_seed(0)\n    x = torch.randn(4, 16, 5)\n\n    print("Shape self-test (all pool x pos combos):")\n    for pool in ("cls", "mean", "last"):\n        for pos in ("learned", "sin"):\n            m = TransformerIntentPredictor(pool=pool, pos=pos)\n            out = m(x)\n            assert out.shape == (4, 1), f"bad shape {out.shape} for pool={pool} pos={pos}"\n    print("  OK — every combo produces (B,1) logits")\n\n    print("\\nParameter ladder (Stage-A sizes; brackets the 594,561-param BiLSTM):")\n    print(f"{\'(d_model, ff)\':>16} {\'L\':>3} {\'params\':>10}")\n    for d_model, ff in [(64, 128), (128, 256), (128, 512)]:\n        for L in (2, 4):\n            n = count_params(TransformerIntentPredictor(d_model=d_model, dim_ff=ff, num_layers=L))\n            print(f"{f\'({d_model}, {ff})\':>16} {L:>3} {n:>10,}")\n\n    default = TransformerIntentPredictor(d_model=128, num_layers=2, dim_ff=256, dropout=0.1,\n                                         pool="cls", pos="learned")\n    print(f"\\ntransformer_default params: {count_params(default):,}  (BiLSTM baseline: 594,561)")\n\n    floor = TransformerIntentPredictor(num_layers=0, pool="mean")\n    assert floor(x).shape == (4, 1)\n    print(f"L=0 linear-probe floor params: {count_params(floor):,}")\n'
(HERE_DIR / "00_transformer_model.py").write_text(model_code)
print("wrote", HERE_DIR / "00_transformer_model.py", f"({len(model_code)} bytes)")
```

```python
# --- output layout + the 2-GPU subprocess runner (same subprocess-per-GPU pattern as
# phase2_kaggle_search/03_search_kaggle.ipynb; jobs here carry their own absolute
# out_dir/out_path, so unlike the search worker, this one needs no --out_root arg). ---
import os, subprocess, sys, tempfile, time

OUT_ROOT = HERE_DIR / "phase4_out"
(OUT_ROOT / "runs_final").mkdir(parents=True, exist_ok=True)
(OUT_ROOT / "runs_loso").mkdir(parents=True, exist_ok=True)
print("OUT_ROOT:", OUT_ROOT)


def run_stage_parallel(jobs, gpu_ids, label=""):
    shards = [[] for _ in gpu_ids]
    for i, job in enumerate(jobs):
        shards[i % len(gpu_ids)].append(job)

    procs, shard_files = [], []
    t0 = time.time()
    for gpu, shard in zip(gpu_ids, shards):
        if not shard:
            continue
        f = tempfile.NamedTemporaryFile(mode="w", suffix=".json", delete=False, dir=str(HERE_DIR))
        json.dump(shard, f)
        f.close()
        shard_files.append(f.name)
        env = dict(os.environ)
        if gpu is not None:
            env["CUDA_VISIBLE_DEVICES"] = str(gpu)
        proc = subprocess.Popen(
            [sys.executable, str(HERE_DIR / "worker_final.py"),
             "--jobs", f.name, "--seq_dir", str(SEQ_DIR), "--here", str(HERE_DIR)],
            env=env,
        )
        procs.append(proc)
        print(f"[{label}] launched worker on gpu={gpu} with {len(shard)} job(s)")

    for p in procs:
        ret = p.wait()
        if ret != 0:
            raise RuntimeError(f"[{label}] worker exited with code {ret}")
    for fn in shard_files:
        os.unlink(fn)
    print(f"[{label}] all workers done in {time.time() - t0:.0f}s")
```

```python
# Stage-C winner, verified 2026-07-11 directly from its raw seed42.json (not
# hand-parsed from the cfg_id string) -- see phase3_search_review/03_search_summary.md.
# val AUC (5-seed) 0.9789 +/- 0.0038 vs transformer_default's 0.9629 +/- 0.0056.
WINNER_CFG = dict(
    d_model=128, nhead=4, num_layers=4, dim_ff=512, dropout=0.1,
    pool="last", pos="sin",
    lr=1e-3, schedule="plateau", weight_decay=1e-5, optimizer="adam",
)
DEFAULT_CFG = engine.DEFAULT_CFG  # transformer_default -- imported, not hand-copied
CONFIGS = {"transformer_searched": WINNER_CFG, "transformer_default": DEFAULT_CFG}
SEEDS = [42, 0, 1, 2, 3]

print("transformer_searched (winner):", WINNER_CFG)
print("transformer_default          :", DEFAULT_CFG)
assert WINNER_CFG != DEFAULT_CFG, (
    "winner should differ from default -- otherwise the search picked nothing new")
```

```python
print("=== Stage D: transformer_searched + transformer_default, 5 seeds each, eval_test=True ===")
print("*** THIS IS THE ONLY PLACE test set03 IS EVALUATED IN THE ENTIRE PROJECT. ***")

stage_d_jobs = []
for tag, cfg in CONFIGS.items():
    for seed in SEEDS:
        out_dir = OUT_ROOT / "runs_final" / tag / f"seed{seed}"
        stage_d_jobs.append(dict(kind="final", cfg=cfg, seed=seed, tag=tag, out_dir=str(out_dir)))
assert len(stage_d_jobs) == 10, len(stage_d_jobs)

run_stage_parallel(stage_d_jobs, GPU_IDS, label="Stage D")


def load_final(job):
    final = json.loads((Path(job["out_dir"]) / "final.json").read_text())
    return {**job, **final}


stage_d_results = [load_final(j) for j in stage_d_jobs]
df_d = pd.DataFrame([
    {"tag": r["tag"], "seed": r["seed"], "val_auc": r["val"]["auc"], "test_auc": r["test"]["auc"]}
    for r in stage_d_results
])
print(df_d.to_string(index=False))

stage_d_summary = []
for tag in CONFIGS:
    sub = df_d[df_d.tag == tag]
    stage_d_summary.append(dict(
        tag=tag, n=len(sub),
        val_auc_mean=float(sub.val_auc.mean()), val_auc_std=float(sub.val_auc.std(ddof=1)),
        test_auc_mean=float(sub.test_auc.mean()), test_auc_std=float(sub.test_auc.std(ddof=1)),
    ))
df_d_summary = pd.DataFrame(stage_d_summary)
print("\n5-seed summary:")
print(df_d_summary.to_string(index=False))
```

```python
print("\n=== Determinism-of-record rerun: transformer_searched seed42, independent re-run ===")
primary_job = next(j for j in stage_d_jobs if j["tag"] == "transformer_searched" and j["seed"] == 42)
primary_final = json.loads((Path(primary_job["out_dir"]) / "final.json").read_text())

rerun_out_dir = OUT_ROOT / "runs_final" / "_determinism_check" / "transformer_searched_seed42_rerun"
rerun_job = dict(kind="final", cfg=WINNER_CFG, seed=42, tag="transformer_searched_rerun",
                 out_dir=str(rerun_out_dir))
run_stage_parallel([rerun_job], GPU_IDS, label="Determinism rerun")
rerun_final = json.loads((rerun_out_dir / "final.json").read_text())

val_delta = abs(rerun_final["val"]["auc"] - primary_final["val"]["auc"])
test_delta = abs(rerun_final["test"]["auc"] - primary_final["test"]["auc"])
print(f"primary  val_auc={primary_final['val']['auc']:.6f}  test_auc={primary_final['test']['auc']:.6f}")
print(f"rerun    val_auc={rerun_final['val']['auc']:.6f}  test_auc={rerun_final['test']['auc']:.6f}")
print(f"|delta|  val={val_delta:.2e}  test={test_delta:.2e}")

determinism_ok = val_delta < 1e-6 and test_delta < 1e-6
print("DETERMINISM CHECK:", "PASS" if determinism_ok
     else "WARN -- did not reproduce exactly; investigate before trusting the headline number")
```
